##### Why this study is interesting: 

In almost all environments, sparse networks found by pruning achieve significantly higher rewards
than the dense baseline. (They train rain up to 80-90% sparse networks with minimal loss in performance compared to the standard dense networks.) 
However training these sparse networks from scratch (static) performs poorly. ("DST algorithms (RigL and SET) improve over static significantly,
however often fall short of matching the pruning performance.") --> leaves room for exploration of transfer benefit in continual RL scenario

They claim that pre-pruned networks don't generalize well in the RL setting as the non-stationarity of the data make it less clear that any decision made at iteration 0 will remain optimal throughout training. 
They compare different, static and dynamic, pruning-techniques with standard dense networks and smaller dense networks by scaling down layer widths to approximately match parameter counts of sparse networks, thereby providing ”parameter-equivalent” dense baseline. --> exactly what my idea also was with the topologies

For the actor-critic methods they claim that the distribution of parameters among actor & critic as well as among different layers, impact training greatly. They observe the best performance by allocating the majority of parameters to the critic network and using Erdos Renyi Kernel (ERK) sparsity distributions --> typically used as random network generator.

They observe robust performance over various hyperparameter variations --> Pretty good to know in advance.

When it comes ot where sparsity is applied in the network, they find that random pruning across layers yields better results, as well as maintaining more density in input and output layers. (For better uhnderstanding: they used Nature CNN and Impala ResNet)
They find improvements across pruning teqniques after adding small weight-decay.

Robustness to Noise: They observe that (1) smaller models are generally more robust to high noise than larger models, (2) sparse models are more robust to high noise than dense models on average,
and (3) in most cases there are minimal differences when the noise is low. Preliminary analysis, suggests that sparse training can produce networks that are more robust to observational noise, even when experienced post-training.

**Challenges**: 
- Observations varying across specific tasks indicate that while sparse training can
bring very significant efficiency gains in some environments, it is not a guaranteed benefit. Unlike in supervised learning, expected gains likely depend on both task and network, and
merits further inquiry.
- Clarity of the differences between sparse and dense training is affected by the stability
of the underlying RL algorithm. --> PPO had too high variance, which is why they used SAC and DQN when studying the different aspects of sparse agents in the rest of their work.
- Initialization seems a little tricky, since depending on how sparsity is added and how pruning happens there are differences in effects of sparsity-aware initialization techniques here

##### Implications for my work: 

Following their structure implies starting with a network and instead of just pruning "small-worldifying" it over time with sparsification and adding long-range connections, similar to their scheme.
At the same time compare this to completely dense networks and ones that are small-world from the beginning (the latter resembles what they call *static* and shows best performance, but needs longer after cold-start). 

The intention behind this is, after [1] proving underexplored potential for sparser networks in DRL, comparing small-world connectivity to dense and to random sparse connectivity and observing influence of connectivity/structure. 


[1] Graesser, L., Evci, U., Elsen, E., & Castro, P. S. (2022, June). The state of sparse training in deep reinforcement learning. In International Conference on Machine Learning (pp. 7766-7792). PMLR.
